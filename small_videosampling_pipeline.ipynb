```python
import cv2
import base64

video_path = "test_video.mp4"

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)

sample_interval = int(fps)

processed_frames = []

frame_number = 0

while True:

    success, frame = cap.read()

    if not success:
        break

    if frame_number % sample_interval == 0:

        timestamp = frame_number / fps

        # Resize image
        resized = cv2.resize(frame, (640, 360))

        # Encode as JPEG
        success, encoded = cv2.imencode(".jpg", resized)

        if success:

            # Convert to Base64
            image_base64 = base64.b64encode(
                encoded
            ).decode("utf-8")

            processed_frames.append({
                "timestamp": timestamp,
                "frame_number": frame_number,
                "image_base64": image_base64
            })

    frame_number += 1

cap.release()

print("Processing completed!")
print("Total processed frames:", len(processed_frames))
print("Sample processed frame details:")
for frame_info in processed_frames[:5]:  # Display first 5 frames
    print(f"Frame Number: {frame_info['frame_number']}, Timestamp: {frame_info['timestamp']:.2f} seconds",base64.b64decode(frame_info['image_base64'])[:10])  # Display first 10 bytes of the image data
```
